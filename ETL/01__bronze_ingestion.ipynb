```python
%sql
select * from read_files(
    '/Volumes/shop_stream/core/raw/customers.csv',
    format => 'csv',
    header => true
)
limit 10
```

```python
%sql
create table if not exists shop_stream.core.customers;

copy into shop_stream.core.customers
from '/Volumes/shop_stream/core/raw/customers.csv'
FILEFORMAT = CSV
FORMAT_OPTIONS ('header' = 'true', 'inferSchema' = 'true', 'mergeSchema' = 'true')
COPY_OPTIONS ('mergeSchema' = 'true')
```
