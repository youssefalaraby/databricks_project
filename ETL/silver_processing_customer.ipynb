```python
from pyspark.sql.functions import *
```

```python
df=spark.table('shop_stream.core.customers')
df.show(10)
```

```python
df.printSchema()
```

```python
df = df.withColumn("name", trim(col("name")))
df.display()
```

```python
df = df.withColumn("city", trim(col("city")))
df = df.withColumn("signup_channel", trim(col("signup_channel")))
df = df.withColumn("country", trim(col("country")))

```

```python
# Check distinct values for city, country, and signup_channel
print("Distinct cities:")
df.select("city").distinct().show(truncate=False)

print("\nDistinct countries:")
df.select("country").distinct().show(truncate=False)

print("\nDistinct signup channels:")
df.select("signup_channel").distinct().show(truncate=False)
```

```python
# Check for duplicate customer_id values
df.groupBy("customer_id").count().filter(col("count") > 1).show()
```

```python
# Check for null values in each column
df.select([sum(col(c).isNull().cast("int")).alias(c) for c in df.columns]).show()
```

```python
df.write.mode("overwrite").format('delta').saveAsTable("shop_stream.silver.customers")
```
