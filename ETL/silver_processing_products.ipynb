```python
from pyspark.sql.functions import *
```

```python
df=spark.table('shop_stream.core.products')
df.display()
```

```python
df = df.withColumn("product_name", trim(col("product_name"))) \
       .withColumn("category", trim(col("category")))
```

```python
df.select("category").distinct().display()
```

```python
df.groupBy("product_id").count().filter(col("count") > 1).display()
```

```python
# Check for null values in each column
df.select([count(when(col(c).isNull(), c)).alias(c) for c in df.columns]).display()
```

```python
# Check for negative values in unit_price and unit_cost columns
df.filter((col("unit_price") < 0) | (col("unit_cost") < 0)).display()
```

```python
df.write.mode("overwrite").format('delta').saveAsTable("shop_stream.silver.products")
```
